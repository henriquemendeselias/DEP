```python
import pandas as pd
import numpy as np
```

```python
#series: é formatado em coluna, diferentemente dos arrays
s = pd.Series([1,2,3, 'ola',6,8])
s
```

```python
#series apartir de uma lista
lista = [10,20,30,40,50]
```

```python
series1 = pd.Series(lista)
```

```python
#series apartir de um dict
dicionario = {'telefone': 123, 'telefone2': 456, 'telefone3': 789}
series2 = pd.Series(dicionario)
```

```python
#series a partir de um array 
arr = np.array([10,20,30,40])
series3 = pd.Series(arr)
series3
```

```python
 #DATAFRAMES
# dataframes a partir de um array
arr = np.random.randint(10,55, size = (4,4))
arr
```
