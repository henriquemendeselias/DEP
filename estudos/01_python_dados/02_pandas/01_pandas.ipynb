```python
indices = ['Janeiro', 'Março', 'Maio', 'Julho', 'Agosto', 'Outubro' ,'Dezembro']
colunas = np.arange(1,32)
dados = np.random.randint(492, 2050, size=(7,31))
```

```python
df = pd.DataFrame(data=dados, index=indices, columns=colunas)
df
```

```python
#quanto vendeu no dia 10 dos meses
df[10]
```

tipos 

df[] muda de comportamento, pode procurar linhas ou colunas-> 
df.loc[] é consistente no padrão df.loc[linha,coluna]

df[coluna]
df[[coluna1, coluna2, coluna3]]
df[coluna][linha]
df[linha1: linha2]
df.loc[ind linha, indice coluna] -> df.loc['Julho', 30]
df.iloc[ind num linha, indice coluna]-> df.iloc[3, 30]
df.loc[[indlinha1, indlinha2, indlinha3], [col1, col2, col3]]
